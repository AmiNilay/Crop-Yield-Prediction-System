```python
# This is a Python representation of a Jupyter Notebook.
# You can copy and paste these cells into a .ipynb file.

# --- CELL 1: Imports and Setup ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Set plotting style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 8)

# --- CELL 2: Load the Data ---
# Make sure your notebook is running from the root directory or adjust the path
try:
    df = pd.read_csv('../data/raw/crop_data.csv')
    print("Data loaded successfully!")
    print("Shape of the dataset:", df.shape)
except FileNotFoundError:
    print("Error: 'crop_data.csv' not found. Make sure it is in the 'data/raw' folder.")

# --- CELL 3: Initial Data Exploration ---
if 'df' in locals():
    print("\nFirst 5 rows of the dataset:")
    display(df.head())

    print("\nDataset Info:")
    df.info()

    print("\nDescriptive Statistics:")
    display(df.describe())

# --- CELL 4: Yield Distribution ---
if 'df' in locals():
    plt.title('Distribution of Crop Yield (tonnes per hectare)')
    sns.histplot(df['Yield'], kde=True, bins=15)
    plt.show()
    # The yield seems to be bimodal, which makes sense as we have different crops (e.g., wheat and rice).

# --- CELL 5: Yield by Crop ---
if 'df' in locals():
    fig = px.box(df, x='Crop', y='Yield', color='Crop', title='Crop Yield by Crop Type')
    fig.show()
    # As expected, Wheat has a higher yield than Rice and Maize in this sample dataset.

# --- CELL 6: Rainfall vs. Yield ---
if 'df' in locals():
    fig = px.scatter(df, x='Rainfall', y='Yield', color='Crop',
                     title='Rainfall vs. Crop Yield',
                     labels={'Rainfall': 'Annual Rainfall (mm)', 'Yield': 'Yield (tonnes/hectare)'},
                     hover_data=['State', 'District'])
    fig.show()
    # We can see a positive correlation. More rainfall generally leads to higher yield, up to a point.

# --- CELL 7: Correlation Heatmap ---
if 'df' in locals():
    # Select only numeric columns for correlation
    numeric_df = df.select_dtypes(include=np.number)
    
    plt.figure(figsize=(14, 10))
    sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Correlation Matrix of Numerical Features')
    plt.show()
    # Observations:
    # - Yield has a strong positive correlation with Rainfall, Soil Nutrients (N, P, K), and Area.
    # - Temperature and Humidity also show a positive correlation.
    # - This confirms that these features are good candidates for our prediction model.
```
